```python
from pathlib import Path
import numpy as np
import pandas as pd


# Load the data
data = np.load('./local_data/tlio_golden/110982076486017/imu0_resampled.npy')
data_raw = np.loadtxt('./local_data/tlio_golden/110982076486017/imu_samples_0.csv',delimiter=',')
print('raw data len:',len(data_raw))
print('calibrated data len:',len(data))
print('resampling size:',len(data_raw)/len(data))
df = pd.DataFrame(data)
df_raw= pd.DataFrame(data_raw)
```

```python
import plotly.graph_objects as go

def compare_columns(col_df, col_df_raw):
    """
    Compare a column from df to a column from df_raw using Plotly.
    
    Parameters:
    col_df (int): Column index from df
    col_df_raw (int): Column index from df_raw
    """
    fig = go.Figure()
    
    # Add trace for df column
    fig.add_trace(go.Scatter(
        x=df[0],
        y=df[col_df],
        mode='lines',
        name=f'Calibrated (Column {col_df})',
        line=dict(color='blue')
    ))
    
    # Add trace for df_raw column
    fig.add_trace(go.Scatter(
        x=df_raw[0]/1000,
        y=df_raw[col_df_raw],
        mode='lines',
        name=f'Raw (Column {col_df_raw})',
        line=dict(color='red')
    ))
    
    fig.update_layout(
        title=f'Column {col_df} (Calibrated) vs Column {col_df_raw} (Raw)',
        xaxis_title='Index (Column 0)',
        yaxis_title='Value',
        hovermode='x unified'
    )
    
    fig.show()
```

```python
compare_columns(6, 6)
```
